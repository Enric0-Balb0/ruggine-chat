```python
import matplotlib.pyplot as plt
import pandas as pd
import matplotlib.dates as mdates
```

```python
# Dati benchmark creazione messaggi
casi = [
    (10, 10, 1),
    (10, 10, 10),
    (100, 10, 10),
    (100, 10, 100),
    (1000, 10, 10),
]

tempo_medio_ms = [26.098797, 5.639972, 4.362299, 2.739854, 3.73231]
msg_s = [38.32, 177.31, 229.24, 364.98, 267.93]

# Etichette leggibili
etichette = [f"G={g}, U={u}, M={m}" for g, u, m in casi]

# Grafico
fig, ax1 = plt.subplots(figsize=(10,6))

# Primo asse: tempo medio per messaggio
ax1.plot(etichette, tempo_medio_ms, marker="o", linewidth=2, label="Tempo medio per messaggio (ms)", color="blue")
ax1.set_ylabel("Tempo medio per messaggio (ms)", color="blue")
ax1.tick_params(axis='y', labelcolor="blue")

# Secondo asse: Msg/s
ax2 = ax1.twinx()
ax2.plot(etichette, msg_s, marker="s", linewidth=2, label="Msg/s", color="orange")
ax2.set_ylabel("Messaggi/s", color="orange")
ax2.tick_params(axis='y', labelcolor="orange")

# Titolo e griglia
plt.title("Benchmark Creazione Messaggi")
ax1.grid(True, linestyle="--", alpha=0.6)

# Legenda combinata
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc="best")

# Etichette X ruotate e layout corretto
ax1.set_xticks(range(len(etichette)))
ax1.set_xlabel("Configurazione (Gruppi, Utenti per gruppo, Messaggi per utente)")

fig.tight_layout()
plt.show()
```
